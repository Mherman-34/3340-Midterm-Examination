Part 5: List Comprehensions

Question 17: Curve the Grades (Loop Version)

```python
scores = [70, 82, 64, 91]

curved = []

for score in scores:
    curved.append(score + 5)

print(curved)
```

Question 18: Curve the Grades

(Comprehension Version)

```python
scores = [70, 82, 64, 91]

curved_2 = [score + 5 for score in scores]

print(curved_2)
```

Question 19: Honor Roll

Part 6: Putting It All Together

Question 20: Quiz Report

```python
sorted_scores = [98, 91, 85, 77, 70, 64, 52]

# Step 1: Get the top three scores
top_three = sorted_scores[:3]

# Step 2: Calculate the average
total = 0

for score in sorted_scores:
    total += score

average = total / len(sorted_scores)

# Step 3: Find passing scores
passed = [score for score in sorted_scores if score >= 70]

# Step 4: Create a tuple
summary = (average, len(passed))

print(top_three)
print(average)
print(passed)
print(summary)
```
