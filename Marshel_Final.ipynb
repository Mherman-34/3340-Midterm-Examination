Stage 1: The Menu

Milestone 1: The Data

```python
# Name of my cafe
cafe_name = 'Home Run Cafe'

# Menu items and prices
menu = {
    'Latte': 4.50,
    'Muffin': 3.25,
    'Bagel': 2.50,
    'Drip Coffee': 2.75,
    'Cookie': 1.75,
    'Iced Tea': 3.00
}

# Cafe operating hours
hours = ('Mon-Fri', 7, 18)

# Allergens for each menu item
allergens = {
    'Latte': {'milk'},
    'Muffin': {'wheat', 'egg'},
    'Bagel': {'wheat'},
    'Drip Coffee': set(),
    'Cookie': {'wheat', 'egg', 'milk'},
    'Iced Tea': set()
}

order = []
sales_log = []

customers = {'Ana': 120, 'Ben': 40}
```

Milestone 2: Show the Menu

```python
# Displays every menu item and its price.
def show_menu(menu):
    print('--- MENU ---')
    for item, price in menu.items():
        print(f'{item}: ${price:.2f}')

show_menu(menu)
```

Milestone 3: Look Up a Price

```python
# Returns an item's price or None if it is unavailable.
def get_price(menu, item):
    try:
        return menu[item]
    except KeyError:
        return None
```

Stage 2: The Order

Milestone 4: Add an Item

```python
# Adds an item if it exists on the menu.
def add_to_order(order, menu, item):
    if item in menu:
        order.append(item)
        return True
    else:
        return False
```

Milestone 5: Remove an Item

```python
# Removes one copy of an item from the order.
def remove_from_order(order, item):
    if item in order:
        order.remove(item)
        return True
    else:
        return False
```

Milestone 6: Work Out the Total

```python
# Calculates the order total including sales tax.
def order_total(order, menu, tax_rate=0.08):
    subtotal = 0
    for item in order:
        subtotal += get_price(menu, item)
    total = subtotal * (1 + tax_rate)
    return round(total, 2)
```

Milestone 7: Menu Categories

```python
categories = {
    'Drinks': ['Latte', 'Drip Coffee', 'Iced Tea'],
    'Food': ['Muffin', 'Bagel', 'Cookie']
}

# Returns the items in a selected category.
def items_in_category(categories, name):
    return categories.get(name, [])

# Displays menu items organized by category.
def show_menu_by_category(menu, categories):
    for category, items in categories.items():
        print(f'--- {category} ---')
        for item in items:
            print(f'{item}: ${menu[item]:.2f}')

show_menu_by_category(menu, categories)
```

Milestone 8: Find Items Under a Price

```python
# Returns items that cost the price limit or less.
def items_under(menu, limit):
    result = []
    for item, price in menu.items():
        if price <= limit:
            result.append(item)
    return result
```

Stage 3: The Customer

Milestone 9: Check for Allergies

```python
# Finds allergens shared by the item and customer.
def allergy_check(item, my_allergies):
    return allergens[item] & my_allergies
```

Milestone 10: Allergy-Safe Items

```python
# Returns items without the customer's listed allergens.
def safe_items(menu, my_allergies):
    safe = []
    for item in menu:
        if len(allergy_check(item, my_allergies)) == 0:
            safe.append(item)
    return safe
```

Milestone 11: Read Customer Input

```python
# Removes extra spaces and capitalizes each word.
def clean_name(text):
    return text.strip().title()

# Checks whether the quantity is a positive integer.
def parse_quantity(text):
    try:
        number = int(text)
    except ValueError:
        return None
    if number > 0:
        return number
    else:
        return None

# Keeps asking until a valid quantity is entered.
def ask_quantity():
    while True:
        text = input('How many? ')
        quantity = parse_quantity(text)
        if quantity is not None:
            return quantity
        print('Please type a whole number greater than 0.')
```

Milestone 12: Loyalty Points

```python
# Adds earned loyalty points to a customer's account.
def add_points(customers, name, total):
    points = int(total)
    if name in customers:
        customers[name] += points
    else:
        customers[name] = points
    return customers[name]

# Determines the customer's loyalty tier.
def get_tier(points):
    if points >= 200:
        return 'Gold'
    elif points >= 100:
        return 'Silver'
    else:
        return 'Bronze'
```

Stage 4: The Receipt and Report

Milestone 13: Print a Receipt

```python
# Counts how many times each item appears in an order.
def count_items(order):
    counts = {}
    for item in order:
        if item in counts:
            counts[item] += 1
        else:
            counts[item] = 1
    return counts

# Prints the receipt, total, and operating hours.
def print_receipt(order, menu):
    print('--- RECEIPT ---')
    counts = count_items(order)
    for item, quantity in counts.items():
        line_total = get_price(menu, item) * quantity
        print(f'{quantity} x {item}: ${line_total:.2f}')
    print(f'Total with tax: ${order_total(order, menu):.2f}')
    days, open_hour, close_hour = hours
    print(f'Open {days}, {open_hour} to {close_hour}')

print_receipt(['Latte', 'Muffin', 'Latte'], menu)
```

Milestone 14: Daily Sales Report

```python
# Calculates daily orders, items sold, and revenue.
def daily_report(sales_log, menu):
    items_sold = 0
    revenue = 0
    unique = set()
    for sale in sales_log:
        items_sold += len(sale)
        revenue += order_total(sale, menu)
        for item in sale:
            unique.add(item)
    return {
        'orders': len(sales_log),
        'items': items_sold,
        'revenue': round(revenue, 2),
        'unique_items': unique
    }

# Displays the daily sales report.
def print_report(report):
    print('--- DAILY REPORT ---')
    print('Orders:', report['orders'])
    print('Items sold:', report['items'])
    revenue = report['revenue']
    print(f'Revenue: ${revenue:.2f}')
    print('Different items sold:', len(report['unique_items']))

test_log = [['Latte', 'Muffin'], ['Cookie'], ['Latte']]
print_report(daily_report(test_log, menu))
```

Stage 5: The Whole Cafe

Milestone 15: Main Program

```python
# Runs the complete cafe ordering system.
def run_cafe():
    my_order = []
    print(f'Welcome to {cafe_name}!')
    customer = clean_name(input('Your name: '))

    while True:
        print()
        print('1) Menu  2) Menu by category  3) Add item  4) Remove item')
        print('5) View order  6) Allergy-safe items  7) Check out  8) Quit')
        choice = input('Choose 1 to 8: ')

        if choice == '1':
            show_menu(menu)

        elif choice == '2':
            show_menu_by_category(menu, categories)

        elif choice == '3':
            name = clean_name(input('Item name: '))
            if get_price(menu, name) is None:
                print('Sorry, we do not have that item.')
            else:
                quantity = ask_quantity()
                for _ in range(quantity):
                    add_to_order(my_order, menu, name)
                print(f'Added {quantity} x {name}.')

        elif choice == '4':
            name = clean_name(input('Item to remove: '))
            if remove_from_order(my_order, name):
                print(f'Removed one {name}.')
            else:
                print('That item is not in your order.')

        elif choice == '5':
            print('Your order:', my_order)

        elif choice == '6':
            allergy_text = input('Your allergies, separated by commas: ')
            my_allergies = set()
            for word in allergy_text.split(','):
                my_allergies.add(word.strip().lower())
            print('Safe for you:', safe_items(menu, my_allergies))

        elif choice == '7':
            print_receipt(my_order, menu)
            points = add_points(customers, customer, order_total(my_order, menu))
            print(f'{customer} now has {points} points ({get_tier(points)}).')
            sales_log.append(my_order)
            my_order = []

        elif choice == '8':
            print('Goodbye!')
            print_report(daily_report(sales_log, menu))
            break

        else:
            print('Please choose a number from 1 to 8.')
```

```python
run_cafe()
```

Extension B: Best Seller

```python
# Returns the most frequently ordered item.
def best_seller(order):
    counts = count_items(order)
    best_item = None
    best_count = 0
    for item, quantity in counts.items():
        if quantity > best_count:
            best_item = item
            best_count = quantity
    return best_item
```
